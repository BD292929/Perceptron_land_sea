```python
w = tk.Tk()

w.title("Land and Sea check")

# שם את החלון באמצע המסך
screen_width = w.winfo_screenwidth()
screen_height = w.winfo_screenheight()

window_width = 1000
window_height = 700

x = (screen_width // 2) - (window_width // 2)
y = (screen_height // 2) - (window_height // 2)

w.geometry(f"{window_width}x{window_height}+{x}+{y}")
w.resizable(True, True)

w.config(background="gray50")

# פונקציית עזר לטעינה ושינוי גודל במקום אחד
def load_image(path, width, height):
    img = Image.open(path)
    img = img.resize((width, height))
    return img

c = [
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea1.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea2.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea3.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea4.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea5.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea6.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea7.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea8.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea9.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\sea\sea10.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land1.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land2.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land3.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land4.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land5.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land6.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land7.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land8.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land9.png',300,300),
    load_image(r'C:\Users\keren\Downloads\land and sea\land\land10.png',300,300)
]


# יצירת שני חצאי מסך
left_frame = tk.Frame(w, bg="gray50")
left_frame.place(relx=0, rely=0, relwidth=0.5, relheight=1)

right_frame = tk.Frame(w, bg="gray50")
right_frame.place(relx=0.5, rely=0, relwidth=0.5, relheight=1)

# תמונות
label1 = tk.Label(left_frame)
label1.place(relx=0.5, rely=0.5, anchor="s")

label2 = tk.Label(right_frame)
label2.place(relx=0.5, rely=0.5, anchor="s")

# כיתובים בתחתית
lb_s = tk.Label(left_frame, text="", font=("Arial", 24))
lb_s.place(relx=0.5, rely=0.55, anchor="center")

lb_l = tk.Label(right_frame, text="", font=("Arial", 24))
lb_l.place(relx=0.5, rely=0.55, anchor="center")


# חיזוי באמצעות TensorFlow
def Predict(img):

    img = img.convert("RGB")

    rgb = np.array(img, dtype=np.float32)

    rgb = np.mean(rgb, axis=(0, 1))

    # נרמול בהתאם לאימון
    rgb = rgb / 255.0

    rgb = np.array([rgb])

    prediction = model.predict(rgb, verbose=0)[0][0]

    if prediction >= 0.5:
        return 1 
    else:
        return 0


# עדכון תמונה, כיתוב וצבע
def update_image(img, image_label, text_label, frame):

    photo = ImageTk.PhotoImage(img)

    image_label.config(image=photo)
    image_label.image = photo

    prediction = Predict(img)

    print("Prediction:", prediction)

    if prediction == 0:
        text_label.config(text="Land", bg="#654321", fg="white")
        frame.config(bg="#654321")

    else:
        text_label.config(text="Sea", bg="#0066CC", fg="white")
        frame.config(bg="#0066CC")


# החלפת תמונות
def switch():

    img1 = random.choice(c)
    img2 = random.choice(c)

    update_image(img1, label1, lb_s, left_frame)
    update_image(img2, label2, lb_l, right_frame)


# תמונות ראשוניות
img1 = random.choice(c)
img2 = random.choice(c)

update_image(img1, label1, lb_s, left_frame)
update_image(img2, label2, lb_l, right_frame)


# כפתור באמצע המסך
button = tk.Button(
    w,
    text="Switch",
    command=switch,
    font=("Arial", 16)
)

button.place(relx=0.5, rely=0.5, anchor="center")

w.mainloop()
```
